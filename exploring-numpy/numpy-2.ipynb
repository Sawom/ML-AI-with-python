```python
import numpy as np

arr = np.random.randint(1 ,100 , size =(10,5)) # sample - 10 , feature - 5
```

```python
b = arr.reshape(5,10)

b.shape
b.ndim
b
```

```python
flatten = b.flatten()
print(flatten.ndim)
flatten

column_wise_flattening = np.ravel(b,order= 'C')

print(column_wise_flattening)

#flatten() এবং ravel() এর পার্থক্য:
#১. মেমরি শেয়ারিং (View বনাম Copy):

#flatten() সবসময় নতুন একটি কপি (Copy) তৈরি করে।
#bbut np.ravel() সাধারণত মূল অ্যারের একটি ভিউ (View) রিটার্ন করে (অর্থাৎ মেমরিতে আলাদা জায়গা নেয় না, মূল অ্যারের সাথে যুক্ত থাকে)। তাই ravel() একটু দ্রুত কাজ করে।
#২. অর্ডার অপশন: ravel()-এ বিভিন্ন অর্ডার ব্যবহার করা যায় (যেমন: order='C' মানে রো-ওয়াইজ, order='F' মানে কলাম-ওয়াইজ ইত্যাদি), যা flatten()-এ নির্দিষ্ট করে দেওয়া যায় না।
```

## arithmetic operator and mathematical functions

```python
# faster for vectorization

x = np.array([ 10 , 8 , 30 , 100 ])
y = np.array([ 2 , 3 ,4  ,5 ])

add = np.add(x,y)
print(add)

sub = x-y
print(sub)

mul = x*y
print(mul)

div = x/y
print(div)

remainder = x % y

print(remainder)
```

```python
# mathematical function

# trigonometry

# sin , cos , tan takes values in radian

sin_val = np.sin(x)
print(sin_val)

cos_val = np.cos(x)
print(cos_val)

deg_con = np.rad2deg(x)
print(deg_con)

# deg2rad
```

```python
#log

x = np.array([ 10 , 8 , 16 , 100 ])
base_10_log_val = np.log10(x)
print(base_10_log_val)


two_diye_bhag = np.log2(x)
print(two_diye_bhag)

sqrt= np.sqrt(x)
print(sqrt)
```

```python
# different
s = np.sum(x)
print(s)

#prefix sum
cumul =np.cumulative_sum(x)
print(cumul)
```

## Logical Function

```python
# comparison - >,< , >= ,<= , == , !=

x = np.array([ 10 , 8 , 16 , 100 ])
y = np.array([ 2 , 3 ,16  ,5 ])

greater_than = x>y

print(greater_than)

equal = x==y

print(equal)

# all () -> shob true kina
# any() -> jekono ekta true ase kina

print(np.all(greater_than))

print(np.any(equal))
```

```python
a = np.random.randint(1 ,100 , size =(100,))
print(a)
```

```python
value_great_than_60 = np.count_nonzero(a==97)
print(value_great_than_60)
# np.count_nonzero() ফাংশনটি দিয়ে কোনো অ্যারের মধ্যে শূন্য (0) ছাড়া বাকি কয়টি উপাদান আছে, অর্থাৎ যেগুলো শূন্য নয় (Non-zero elements), সেই সংখ্যাটি গুনে বের করা হয়।
```

```python
unique_value,count = np.unique(a,return_counts=True)

print(unique_value)
print(count)
```

## statistical function

```python
#data load

data = np.genfromtxt('student_scores.csv' ,delimiter=',',skip_header=1)

print(data)

# np.genfromtxt-এর পূর্ণরূপ হলো "generate from text"।
# এটি NumPy-র একটি অত্যন্ত কাজের ফাংশন, যা কোনো টেক্সট ফাইল (যেমন: .csv বা .txt ফাইল) থেকে ডেটা রিড করে সেগুলোকে একটি NumPy অ্যারেতে রূপান্তর করতে ব্যবহার করা হয়।
```

```python
# statistics
math_marks = data[::,:1]

print(math_marks.T)

max_math_marks = np.max(math_marks)
print(max_math_marks)

min_math_marks = np.min(math_marks)
print(min_math_marks)

average_math_marks = np.mean(math_marks)
print(average_math_marks)

median_math_marks = np.median(math_marks)
print(median_math_marks)

# standard deviation

std_math_marks = np.std(math_marks)
print(std_math_marks)


```
